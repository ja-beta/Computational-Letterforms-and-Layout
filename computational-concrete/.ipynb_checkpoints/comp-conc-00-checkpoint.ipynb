# Custom fonts!

All the automating opportunities last class made me think of all the "annoying" parts of css that one usually has to go through, like importing fonts, so this seems like a nice opportunity to have it generate the "@font-face" css code that I would usually need to manually type by hand. There are probably smarter ways of doing this but I don't like abstracting everything and would like to retain some control so I'm going with a function that just goes through a list of fonts files (that I have just included in my project folder), and generates the appropriate css import rules for each one. 

```python
def get_fonts_css(fonts):
    css = ""
    for name, path, weight, style in fonts:
        with open(path, 'rb') as f:
            font_data = f.read()
        base64_font = base64.b64encode(font_data).decode('utf-8')
        css += f"""
        @font-face {{
            font-family: '{name}';
            src: url(data:font/opentype;base64,{base64_font}) format('opentype');
            font-weight: {weight};
            font-style: {style};
        }}
        """
    return css
```

Fonts are binary data - so if we want to embed it directly in HTML/CSS, we need to convert the binary data into text (like a string of ASCII characters) that can be included in the source code. Base64 encoding is a standardized way of doing it (or so I've learned). It meand it's essentially a string of 64 ASCII characters.
From my experience / understanding this would not be the best approach for actual sites as it drastically increases the weight of the HTML file (as it's embedded in it), but this might be the way to go for our porpuses here, where I'd rather have something reliable that I can easily share.

```python
font_files = [
    ('80-kb-g', './fonts/80-kb-Grunge.otf', '400', 'normal'),
    ('80-kb', './fonts/80-kb-Sharp.otf', '400', 'normal'),
    ('80-kb', './fonts/80-kb-SharpBold.otf', '600', 'bold'),
    ('80-kb', './fonts/80-kb-SharpBoldItalic.otf', '600', 'italic'),
    ('80-kb', './fonts/80-kb-SharpItalic.otf', '400', 'italic'),
    ('80-kb-s', './fonts/80-kb-Soft.otf', '400', 'normal')
]
```

```python
css = get_fonts_css(font_files)
```

```python
html_tmpl = """<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="utf-8">
    <title>{title}</title>
    <style>
    html {{ 
        min-height: 5em; 
        overflow: hidden;
        font-family: '80-kb', monospace;
        font-size: 18px;
        }}

    body{{
        background-color: transparent;
    }}

    h1 {{
        font-weight: 400;
    }}

    div{{
        background-color: transparent;
    }}
    </style>
</head>
<body>
{content}
</body>
</html>"""
```

```python
def mkdiv(content, **kwargs):
    if 'position' not in kwargs:
        kwargs['position'] = 'absolute'
    style_str = ' '.join([": ".join((k.replace('_', '-'), v))+";" for k, v in kwargs.items()])
    return f"<div style='{style_str}'>{content}</div>"
```

```python
words = q0.split()
before_again = ' '.join(words[:-1])
again_w = words[-1]
print(before_again, again_w)
```

```python
elements = []

elements.append(mkdiv(before_again, 
                     position="absolute", 
                     top="78px",
                     left="48%",
                     font_size="12px",
                     transform="translate(-50%, -50%)",
                     z_index="1",
                     color="white"
                     )
               )


repetitions = 80
max_space = 12

for i in range(repetitions):
    if i < repetitions / 20: # for the first x percent (100/x=%), the space increases as i increases
        space_factor = i / (repetitions / 3) * max_space
    else: # for the remaining, the space decreases as i increases
        space_factor = (repetitions - i) / (repetitions / 3) * max_space
    
    spaced_again = ""
    for char in again_w:
        spaced_again += char + "&nbsp;" * int(space_factor)
    
    #calculates the percentage of the way through the repetitions
    t = i / repetitions  
    
    left_pos = 60 
    
    # Parabolic arc - the formula 2*t*(1-t) creates a curve that rises and falls
    top_pos = 72 + 1200 * (2 * t * (1-t))

    opacity = 0.1 + 0.6 * (i / repetitions)
    
    elements.append(mkdiv(
        spaced_again, 
        position="absolute", 
        left=f"{left_pos}%",  
        transform="translateX(-50%)", 
        top=f"{top_pos}px",
        opacity= "1",
        font_size="12px",
        background_color="transparent",
        color="white",
    ))


poem_content = "".join(elements)
container_div = f"""
<div style="position: relative; 
            margin: 10px; 
            min-height: 800px;
            background-color: black;
            
            ">{poem_content}
</div>"""
html_src = html_tmpl.format(title="again", content=container_div)
show_html(html_src)
```

```python
#tried to play around with the idea of multiple columns of the repetitions of this text where the color changes. not loving where I got but keeping this so I can come back to it with new thoughts...

words = q1.split()

def color_gradient(start_rgb, end_rgb, percent):
    r1, g1, b1 = start_rgb
    r2, g2, b2 = end_rgb
    
    r = r1 + percent * (r2 - r1)
    g = g1 + percent * (g2 - g1)
    b = b1 + percent * (b2 - b1)
    
    return f'rgb({int(r)}, {int(g)}, {int(b)})'

def create_block(x_position, height_factor=1.0, start_offset=0):
    tree_elements = []
    
    base_variations = []
    for i in range(len(words)+1):
        rotated_words = words[i:] + words[:i]
        base_variations.append(" ".join(rotated_words))
    
    repetitions = random.randint(1, 3)
    variations = base_variations * repetitions
    
    vertical_spacing = 4 * height_factor
    start_y = 50 + start_offset
    
    top_color = (46, 139, 87) 
    bottom_color = (112, 74, 47)
    
    for idx, sentence in enumerate(variations):
        percent = idx / (len(variations) - 1) if len(variations) > 1 else 0
        color = color_gradient(top_color, bottom_color, percent)
        
        tree_elements.append(mkdiv(
            sentence,
            position="absolute",
            left=f"{x_position}%",
            top=f"{start_y + idx * vertical_spacing}px",
            transform="translateX(-50%)",
            color=color,
            font_size="8px",
            letter_spacing="-0.5px",
            font_family="80-kb-g",
            background_color="transparent"
        ))
    
    return tree_elements

elements = []
num_trees = 10

start_x = 10    
end_x = 90 
x_range = end_x - start_x

for i in range(num_trees):
    x_pos = start_x + (i * x_range / (num_trees - 1))
    
    height = random.uniform(1, 5)
    offset = random.randint(-80, 40)
    
    tree = create_block(x_pos, height, offset)
    elements.extend(tree)

container = """position: relative; 
            margin: 40px; 
            min-height: 500px;
            border-radius: 50%;
            width:500px;
            background-color: #181818;
            margin-left: auto;
            margin-right: auto;
            overflow: hidden"""

poem_content = "".join(elements)
div = f"<div style='{container}'>{poem_content}</div>"
html_src = html_tmpl.format(title="forest of matters", content=div)
show_html(html_src)
```
