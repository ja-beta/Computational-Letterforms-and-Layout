# String interpolation

```python
animals = ["alligator", "babboon", "capybara", "dinosaur", "emu"]
```

```python
print(f"I really love {animals[0]}!")
```

```python
for item in animals:
    message = "I really love " + item + "!"
    print(message)
```

```python
num = list(range(10))
num
```

```python
# I have (n) dinosaurs
for item in num:
    message = "I have " + str(item) + " dinosaurs"
    print(message)
```

```python
import random
for item in animals:
    x = random.randrange(100)
    y = random.randrange(100)
    message = '<div style="top: ' + str(x) + 'px; left: ' + str(y) + 'px">yay ' + item + '</div>'
    print(message)
```

# .format()
every string has a method called format. It will automatically convert numbers to strings for us.

```python
"hello".format()
```

```python
message = "hello {x}!"
message.format(x="world")
```

```python
template = '<div style="top: {x}px; left: {y}px;">{animal} (mass: {kg: .2f} kg)</div>' # the ".2f" will format to 2 decimal places
for item in animals:
    x = random.randrange(100) # for ints
    y = random.randrange(100)
    kg = random.uniform(5, 5000) # for floats
    message = template.format(x=x, y=y, animal=item, kg=kg)
    print(message)
```

```python
html_template = """
<html>
    <head>
        <title>{title}</title>
    </head>

    <body>
        <div>this is an html template for {corp_name}!</div>
    </body>

</html> """
```

```python
html_template.format(title="Welcome to Big Bad Co. Inc. Ltd.",
                     corp_name="Big Bad Big Co. Inc. Ltd.")

print(html_template)
```

# F strings!
relatively new to python. 

```python
greeting = "hello"
location = "world"
message =f"{greeting} {location}!"
print(message)
```

```python
for item in animals:
    x = random.randrange(100) # for ints
    y = random.randrange(100)
    kg = random.uniform(5, 5000) # for floats
    message = f"""
<div style="top: {x}px; left: {y}px;">
    {item.upper()} (mass: {kg: .2f} kg)
</div>
"""

    print(message)
```
